```python
from sklearn import datasets
import pandas as pd
```

```python
d=datasets.load_iris()
d
```

```python
x=d.data
y=d.target
```

```python
x=d.data[:,:2]
y=(d.target==0).astype(int)
```

```python
x
```

```python
y
```

Train, Test, Split

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=52)
```

```python
from sklearn.svm import SVC
```

```python
model=SVC(kernel='linear',C=1)
```

```python
model.fit(x_train,y_train)
```

```python
predictions = model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_test, predictions)
print(f"Model Accuracy: {accuracy * 100:.2f}%")
```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.scatter(x[:,0],x[:,1],c=y)

w=model.coef_[0]
b=model.intercept_[0]

x_points=np.linspace(x[:,0].min(),x[:,0].max(),100)
y_decision=-(w[0]/w[1])*x_points-b/w[1]
y_margin1=-(w[0]/w[1])*x_points-(b-1)/w[1]
y_margin2=-(w[0]/w[1])*x_points-(1+b)/w[1]

plt.plot(x_points,y_decision,'k-')
plt.plot(x_points,y_margin1,'k--')
plt.plot(x_points,y_margin2,'k--')
plt.scatter(model.support_vectors_[:,0],model.support_vectors_[:,1],s=100,facecolors='none',edgecolors='red')

plt.xlabel('Sepel Length')
plt.ylabel('Sepel Width')
plt.title('Linear SVM Decision Boundary')
plt.show()
```
